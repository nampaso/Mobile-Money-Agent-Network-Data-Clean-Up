```python
import os
import pandas as pd
import numpy as np
from generate_data import generate_raw_data
from cleaning_pipeline import clean_agent_data
```

```python
# Generate raw datasets dynamically
txns_raw, agents_raw = generate_raw_data(seed=44)
```

```python
# Execute data cleaning pipeline
clean_df, audit_log = clean_agent_data(txns_raw, agents_raw)
```

```python
# Display Data Quality Audit Table
print("--- DATA QUALITY AUDIT TABLE ---")
    audit_log
```

```python
# Post-Merge Validation Statistics
print("--- POST-MERGE VALIDATION STATISTICS ---")
print(f"Raw Transactions Row Count: {len(txns_raw)}")
print(f"Clean Transactions Row Count: {len(clean_df)}")
unknown_agents_count = (clean_df['county'] == 'Unknown').sum()
print(f"Transactions with Unknown Agents: {unknown_agents_count}")
```

```python
# County Summary Table
county_summary = clean_df.groupby('county').agg(
    total_transactions=('txn_id', 'count'),
    total_amount=('amount', 'sum'),
    average_amount=('amount', 'mean')
).reset_index()

print("--- COUNTY SUMMARY TABLE ---")
county_summary
```

```python
# Weekly Pivot Table
clean_df['txn_date_dt'] = pd.to_datetime(clean_df['txn_date'])
clean_df['week'] = clean_df['txn_date_dt'].dt.isocalendar().week

weekly_pivot = clean_df.pivot_table(
    index='week',
    columns='txn_type',
    values='amount',
    aggfunc=['count', 'sum'],
    fill_value=0
)

print("--- WEEKLY PIVOT TABLE ---")
weekly_pivot
```

```python
# Export cleaned output
os.makedirs('outputs', exist_ok=True)
export_df = clean_df.drop(columns=['txn_date_dt', 'week'])
export_df.to_csv('outputs/agent_transactions_clean.csv', index=False)
print("Saved outputs/agent_transactions_clean.csv successfully!")
```

# One-Page Data Quality Report

## 1. Uniqueness
Duplicate transaction records were detected in the raw transactions data (`3` rows) and the agents registry (`2` rows). All duplicates were deduplicated based on primary keys (`txn_id` and `agent_id`) keeping the initial occurrence to ensure data integrity.

## 2. Completeness
A total of `40` transaction records were missing amounts. Additionally, `1` negative transaction amount was detected and set to missing. All missing values were imputed using the median transaction amount to maintain statistical balance without skewing the distribution. An `amount_imputed` flag column was added for full auditability.

## 3. Consistency
Non-standard transaction type values (such as `dep`, `depo`, `WITHDRAWAL`, and `Withdraw`) were mapped to standard values (`Deposit` and `Withdrawal`). Agent keys were converted to uppercase to eliminate casing discrepancies.

## 4. Timeliness / Validity
Dates were harmonized from heterogeneous strings into ISO 8601 formatting (`YYYY-MM-DD HH:MM:SS`). Invalid transaction amounts (negative values) were invalidated prior to imputation.

## 5. Accuracy
Using the Interquartile Range (IQR) method, `120` extreme transaction amount outliers were flagged. An explicit boolean flag `is_outlier` was included to allow analytical downstream filtering while preserving original data.

## 6. Referential Integrity
Transactions were matched to agent profiles. Unmatched transactions with unknown foreign keys (`agent_id` = `A999`) were safely assigned `Unknown` in county fields during left join resolution.
